```python
# Step 4: Load the orders dataset and take a first look
import pandas as pd

# Load the file
orders = pd.read_csv(f"{data_path}/olist_orders_dataset.csv")

# Basic information
print("ORDERS DATASET")
print(f"  Rows:    {len(orders)}")
print(f"  Columns: {len(orders.columns)}")
print(f"\nColumn names:")
for col in orders.columns:
    print(f"  - {col}")
print(f"\nFirst 3 rows:")
orders.head(3)
```

```python
# Step 5: Load the customers dataset
customers = pd.read_csv(f"{data_path}/olist_customers_dataset.csv")

print("CUSTOMERS DATASET")
print(f"  Rows:    {len(customers)}")
print(f"  Columns: {len(customers.columns)}")
print(f"\nColumn names:")
for col in customers.columns:
    print(f"  - {col}")
print(f"\nFirst 3 rows:")
customers.head(3)
```

```python
# Step 6: Find returning customers
# Join orders with customers to get customer_unique_id per order
orders_with_customers = orders.merge(customers, on='customer_id')

# Count how many times each unique customer appears
purchase_counts = orders_with_customers\
    .groupby('customer_unique_id')['order_id']\
    .count()\
    .reset_index()
purchase_counts.columns = ['customer_unique_id', 'total_orders']

# Label them
purchase_counts['is_returning_customer'] = \
    purchase_counts['total_orders'] > 1

# Summary
returning = purchase_counts['is_returning_customer'].sum()
one_time  = (~purchase_counts['is_returning_customer']).sum()
total     = len(purchase_counts)

print("RETURNING CUSTOMER ANALYSIS")
print(f"  Total unique customers:    {total:,}")
print(f"  Returning customers:       {returning:,} ({returning/total:.1%})")
print(f"  One-time customers:        {one_time:,} ({one_time/total:.1%})")
print(f"\nSample of the data:")
purchase_counts.head(5)
```

```python
# Step 7: Load the reviews dataset
reviews = pd.read_csv(f"{data_path}/olist_order_reviews_dataset.csv")

print("REVIEWS DATASET")
print(f"  Rows:    {len(reviews)}")
print(f"  Columns: {len(reviews.columns)}")
print(f"\nColumn names:")
for col in reviews.columns:
    print(f"  - {col}")

# Look at the score distribution
print(f"\nReview score distribution:")
score_counts = reviews['review_score'].value_counts().sort_index()
for score, count in score_counts.items():
    bar = "█" * int(count/1000)
    print(f"  {score} star: {count:,} reviews  {bar}")
```

```python
# Step 8: Create sentiment labels from review scores
import pandas as pd

def assign_sentiment(score):
    if score >= 4:
        return 'POSITIVE'
    elif score == 3:
        return 'NEUTRAL'
    else:
        return 'NEGATIVE'

reviews['sentiment'] = reviews['review_score'].apply(assign_sentiment)

# Summary
print("SENTIMENT DISTRIBUTION")
sentiment_counts = reviews['sentiment'].value_counts()
total = len(reviews)

for sentiment, count in sentiment_counts.items():
    print(f"  {sentiment:<10}: {count:,} ({count/total:.1%})")

print(f"\nSample reviews with labels:")
reviews[['review_score', 'sentiment',
         'review_comment_message']].dropna().head(5)
```

```python
# Step 9: Load the marketing funnel datasets
leads = pd.read_csv(f"{data_path}/olist_marketing_qualified_leads_dataset.csv")
deals = pd.read_csv(f"{data_path}/olist_closed_deals_dataset.csv")

print("QUALIFIED LEADS DATASET")
print(f"  Rows:    {len(leads)}")
print(f"  Columns: {len(leads.columns)}")
print(f"  Column names: {list(leads.columns)}")

print(f"\nCLOSED DEALS DATASET")
print(f"  Rows:    {len(deals)}")
print(f"  Columns: {len(deals.columns)}")
print(f"  Column names: {list(deals.columns)}")

# How many leads converted?
converted = leads['mql_id'].isin(deals['mql_id']).sum()
not_converted = len(leads) - converted

print(f"\nLEAD CONVERSION ANALYSIS")
print(f"  Total leads:       {len(leads):,}")
print(f"  Converted:         {converted:,} ({converted/len(leads):.1%})")
print(f"  Not converted:     {not_converted:,} ({not_converted/len(leads):.1%})")
```

```python
# Step 10: Save a summary of our data exploration
import json
from datetime import datetime

summary = {
    "exploration_date": datetime.now().strftime("%Y-%m-%d"),
    "datasets": {
        "orders": {"rows": len(orders), "columns": len(orders.columns)},
        "customers": {"rows": len(customers), "columns": len(customers.columns)},
        "reviews": {"rows": len(reviews), "columns": len(reviews.columns)},
        "leads": {"rows": len(leads), "columns": len(leads.columns)},
        "deals": {"rows": len(deals), "columns": len(deals.columns)}
    },
    "key_findings": {
        "total_unique_customers": 96096,
        "returning_customers_pct": 3.1,
        "lead_conversion_pct": 10.5,
        "positive_reviews_pct": 77.1,
        "negative_reviews_pct": 14.7,
        "neutral_reviews_pct": 8.2
    },
    "ground_truth_labels": {
        "W1_customer_behaviour": "is_returning_customer (True/False)",
        "W2_marketing_funnel": "is_converted (True/False)",
        "W3_product_seller": "sentiment (POSITIVE/NEUTRAL/NEGATIVE)"
    }
}

# Save to Results folder
import json
output_path = f"{results_path}/data_exploration_summary.json"
with open(output_path, 'w') as f:
    json.dump(summary, f, indent=2)

print("✓ Data exploration summary saved to Results folder")
print(f"  Location: {output_path}")
print(f"\nKey findings:")
for key, value in summary['key_findings'].items():
    print(f"  {key}: {value}")
```
